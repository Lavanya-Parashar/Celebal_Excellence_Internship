Create Products table

**STEP 2: PERFORMING QUERIES**

```python
# Q1. Find all orders where sales are greater than the average sales (Subquery)

query = """
SELECT *
FROM orders
WHERE Sales > (
    SELECT AVG(Sales)
    FROM orders
);
"""

run_query(query)
```

```python
# Q2. Find the highest sales order for each customer (Subquery)

query = """
SELECT *
FROM orders o
WHERE Sales = (
    SELECT MAX(Sales)
    FROM orders
    WHERE `Customer ID` = o.`Customer ID`
);
"""

run_query(query)
```

```python
# Q3. Calculate total sales for each customer (CTE)

query = """
WITH customer_sales AS (
    SELECT
        `Customer ID`,
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY `Customer ID`
)

SELECT *
FROM customer_sales;
"""

run_query(query)
```

```python
# Q4. Find customers whose total sales are above average (CTE + Subquery)

query = """
WITH customer_sales AS (
    SELECT
        `Customer ID`,
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY `Customer ID`
)

SELECT *
FROM customer_sales
WHERE total_sales > (
    SELECT AVG(total_sales)
    FROM customer_sales
);
"""

run_query(query)
```

```python
# Q5. Rank all customers based on total sales (Window Function)

query = """
WITH customer_sales AS (
    SELECT
        `Customer ID`,
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY `Customer ID`
)

SELECT *,
       RANK() OVER (ORDER BY total_sales DESC) AS customer_rank
FROM customer_sales;
"""

run_query(query)
```

```python
# Q6. Assign row numbers to each order within a customer
# (Window Function + PARTITION BY)

query = """
SELECT *,
       ROW_NUMBER() OVER (
           PARTITION BY `Customer ID`
           ORDER BY Sales DESC
       ) AS row_num
FROM orders;
"""

run_query(query)
```

```python
# Q7. Display top 3 customers based on total sales
# (Window Function)

query = """
WITH customer_sales AS (
    SELECT
        `Customer ID`,
        SUM(Sales) AS total_sales
    FROM orders
    GROUP BY `Customer ID`
)

SELECT *
FROM (
    SELECT *,
           RANK() OVER (ORDER BY total_sales DESC) AS rank_num
    FROM customer_sales
)
WHERE rank_num <= 3;
"""

run_query(query)
```

**FINAL COMBINED QUERY**

```python
# Final Combined Query
# Show Customer Name, Total Sales, and Rank
# Using JOIN + CTE + Window Function

query = """
WITH customer_sales AS (
    SELECT
        c.`Customer Name`,
        SUM(o.Sales) AS total_sales
    FROM customers c
    JOIN orders o
    ON c.`Customer ID` = o.`Customer ID`
    GROUP BY c.`Customer Name`
)

SELECT
    `Customer Name`,
    total_sales,
    RANK() OVER (ORDER BY total_sales DESC) AS customer_rank
FROM customer_sales;
"""

run_query(query)
```

```python
# Q1. Top 5 customers based on total sales

query = """
WITH customer_sales AS (
    SELECT
        c.`Customer Name`,
        SUM(o.Sales) AS total_sales
    FROM customers c
    JOIN orders o
    ON c.`Customer ID` = o.`Customer ID`
    GROUP BY c.`Customer Name`
)

SELECT *
FROM customer_sales
ORDER BY total_sales DESC
LIMIT 5;
"""

run_query(query)
```

```python
# Q2. Bottom 5 customers based on total sales

query = """
WITH customer_sales AS (
    SELECT
        c.`Customer Name`,
        SUM(o.Sales) AS total_sales
    FROM customers c
    JOIN orders o
    ON c.`Customer ID` = o.`Customer ID`
    GROUP BY c.`Customer Name`
)

SELECT *
FROM customer_sales
ORDER BY total_sales ASC
LIMIT 5;
"""

run_query(query)
```

```python
# Q3. Customers who made only one order

query = """
SELECT
    c.`Customer Name`,
    COUNT(o.`Order ID`) AS total_orders
FROM customers c
JOIN orders o
ON c.`Customer ID` = o.`Customer ID`
GROUP BY c.`Customer Name`
HAVING COUNT(o.`Order ID`) = 1;
"""

run_query(query)
```

```python
# Q4. Customers with above-average sales

query = """
WITH customer_sales AS (
    SELECT
        c.`Customer Name`,
        SUM(o.Sales) AS total_sales
    FROM customers c
    JOIN orders o
    ON c.`Customer ID` = o.`Customer ID`
    GROUP BY c.`Customer Name`
)

SELECT *
FROM customer_sales
WHERE total_sales > (
    SELECT AVG(total_sales)
    FROM customer_sales
);
"""

run_query(query)
```

```python
# Q5. Highest order value per customer

query = """
SELECT
    c.`Customer Name`,
    MAX(o.Sales) AS highest_order_value
FROM customers c
JOIN orders o
ON c.`Customer ID` = o.`Customer ID`
GROUP BY c.`Customer Name`;
"""

run_query(query)
```
