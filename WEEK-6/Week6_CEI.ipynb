CREATING SPARK SESSION

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Week6Assignment") \
    .getOrCreate()
```

READ CSV

```python
df = spark.read.csv(
    "fake_retail_data.csv",
    header=True,
    inferSchema=True
)

df.show(5)
df.printSchema()
```

CREATE EXTRA COLUMNS NEEDED FOR WEEK 6

```python
from pyspark.sql.functions import lit, col, when

df = (
    df
    .withColumnRenamed(
        "product_category",
        "category"
    )
    .withColumn(
        "product_id",
        col("user_id")
    )
    .withColumn(
        "amount",
        col("sale_amount")
    )
    .withColumn(
        "base_price",
        col("price")
    )
    .withColumn(
        "priority",
        when(
            col("sale_amount") > 2500,
            "High"
        ).otherwise("Low")
    )
)
```

```python
from pyspark.sql.functions import col, when

df = (
    df
    .withColumnRenamed("product_category", "category")
    .withColumn("product_id", col("user_id"))
    .withColumn("amount", col("sale_amount"))
    .withColumn("base_price", col("price"))
    .withColumn(
        "priority",
        when(col("sale_amount") > 2500, "High")
        .otherwise("Low")
    )
    .withColumn(
        "status",
        when(col("sale_amount") > 1000, "Completed")
        .otherwise("Pending")
    )
)

df.printSchema()
```

**Q5**

```python
from pyspark.sql.functions import col

df = (
    df
    .withColumnRenamed(
        "product_category",
        "category"
    )
    .withColumn(
        "product_id",
        col("user_id")
    )
)
```

```python
df.filter(
    col("category") == "Electronics"
).select(
    "product_id",
    "price"
).show()
```

**Q6**

```python
from pyspark.sql.types import DoubleType

df_new = (
    df.withColumnRenamed(
        "category",
        "new_name"
    )
    .withColumn(
        "price",
        col("price").cast(DoubleType())
    )
)

df_new.printSchema()
```

**Q8**

```python
from pyspark.sql.functions import when, col

df_orders = df.withColumn(
    "status",
    when(col("sale_amount") > 1000, "Completed")
    .otherwise("Pending")
)
```

```python
df_orders.filter(
    (col("status") == "Completed") &
    (col("sale_amount") > 1000)
).show()
```

```python
from pyspark.sql.functions import col
from pyspark.sql.types import DoubleType

df = df.withColumn(
    "base_price",
    col("base_price").cast(DoubleType())
)

df = df.withColumn(
    "final_price",
    col("base_price") * 1.18
)

df.select(
    "base_price",
    "final_price"
).show(10)
```

```python
from pyspark.sql.functions import col, when

df = (
    df
    .withColumn("amount", col("sale_amount"))
    .withColumn("base_price", col("price"))
    .withColumn(
        "priority",
        when(col("sale_amount") > 2500, "High")
        .otherwise("Low")
    )
)
```

**Q11**

```python
from pyspark.sql.functions import col

df_parquet = spark.read.parquet(
    "input_parquet"
)

clean_df = df_parquet.filter(
    col("user_id").isNotNull()
)

clean_df.write.mode("overwrite").csv(
    "output_csv",
    header=True
)

clean_df.show(5)
```
